Completes the following steps of small task:
Mention Marks of English , Islamiat and Maths out of 100 in 3 different variables
Mention Variable of Total Marks and assign 300 to it
Calculate Percentage

```python
maths = 75
english = 80
islamiat = 88

total_marks = 300

get_marks = maths + english + islamiat

get_percentage = get_marks / total_marks * 100


print(f"Marks in Maths: {maths}")
print(f"Marks in English: {english}")
print(f"Marks in Islamiat: {islamiat}")
print(f"Total Marks: {total_marks}")
print(f"Marks Obtained: {get_marks}")
print(f"Percentage: {get_percentage}%")
```

Take the radius of a circle from a variable and calculate its Area and Circumference (use pi = 3.14159)

```python
radius = 10
pi = 3.14159

area = pi * radius ** 2
circumference = 2 * pi * radius

print(f"Radius: {radius}")
print(f"Area: {area}")
print(f"Circumference: {circumference}")
```

Take the price of an item and a discount percentage in variables, then calculate and print the discount amount and the final price after discount


```python
price = 599
discount_percentage = 10

discount_amount = price * discount_percentage / 100
final_price = price - discount_amount

print(f"Price: {price}")
print(f"Discount Percentage: {discount_percentage}%")
print(f"Discount Amount: {discount_amount}")
print(f"Final Price After Discount: {final_price}")
```
