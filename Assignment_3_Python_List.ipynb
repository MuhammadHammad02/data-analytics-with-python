Write a program that accepts a list from user. Your program should reverse the content of list and
display it. Do not use reverse() method.

```python
# 1. Accept list items from the user (separated by spaces or commas)
user_input = input("Enter elements of the list (separated by space or comma): ")

# 2. Clean the input and convert it into a list
user_list = user_input.replace(",", " ").split()

# 3. Reverse the list without using .reverse() method
# [::-1] means start from the end and move backwards with a step of -1
reversed_list = user_list[::-1]

# 4. Display the results
print("Original List:", user_list)
print("Reversed List:", reversed_list)
```

Find and display the largest number of a list without using built-in function max(). Your program
should ask the user to input values in list from keyboard.


```python
# 1. Accept list items from the user
user_input = input("Enter numbers (separated by space or comma): ")

# 2. Clean input and convert the strings into a list of actual integers
string_list = user_input.replace(",", " ").split()
numbers_list = [int(num) for num in string_list]

# 3. Logic to find the largest number without max()
# Assume the first number is the largest to start with
largest = numbers_list[0]

# Loop through the list to compare each number
for num in numbers_list:
    if num > largest:
        largest = num  # Update largest if a bigger number is found

# 4. Display the result
print("Your numbers:", numbers_list)
print("The largest number is:", largest)
```

. Write a program that rotates the element of a list so that the element at the first index moves to the
second index, the element in the second index moves to the third index, etc., and the element in the last
index moves to the first index.


```python
# 1. Accept list items from the user
user_input = input("Enter elements of the list (separated by space or comma): ")

# 2. Clean input and convert it into a list
user_list = user_input.replace(",", " ").split()

# Ensure the list is not empty before rotating
if user_list:
    # 3. Rotate elements using list slicing
    # user_list[-1:] gets the last element as a list
    # user_list[:-1] gets all elements except the last one
    rotated_list = user_list[-1:] + user_list[:-1]

    # 4. Display the results
    print("Original List:", user_list)
    print("Rotated List: ", rotated_list)
else:
    print("The list is empty.")
```

Write a program that input a string and ask user to delete a given word from a string.

```python
import re

main_string = input("Enter a sentence: ")
word_to_delete = input("Enter the word you want to delete: ")

# Using Regular Expressions (re) to ignore case differences flags=re.IGNORECASE
modified = re.sub(rf'\b{word_to_delete}\b', '', main_string, flags=re.IGNORECASE)
cleaned_string = " ".join(modified.split())

print("Modified Sentence:", cleaned_string)
```

 Write a program with a function that accepts a string from keyboard and create a new string after
converting character of each word capitalized. For instance, if the sentence is "stop and smell the roses."
the output should be "Stop And Smell The Roses

```python
# 1. Define the capitalization function
def capitalize_words(sentence):
    # Split the sentence into individual words
    words = sentence.split()

    # Capitalize the first letter of each word using a list comprehension
    # .capitalize() ensures the first letter is uppercase and the rest are lowercase
    capitalized_words = [word.capitalize() for word in words]

    # Join the words back together with a single space
    new_string = " ".join(capitalized_words)
    return new_string

# 2. Accept input from the keyboard
user_input = input("Enter a sentence: ")

# 3. Call the function and store the result
result = capitalize_words(user_input)

# 4. Display the original and modified strings
print("\nOriginal Sentence:", user_input)
print("Modified Sentence:", result)
```

. Find the sum of each row of matrix of size m x n. For example for the following matrix output will be
like this :
Sum of row 1 = 32
Sum of row 2 = 31
Sum of row 3 = 63

```python
# 1. Get matrix dimensions from the user
m = int(input("Enter the number of rows (m): "))
n = int(input("Enter the number of columns (n): "))

matrix = []

# 2. Input the matrix values row by row
print(f"\nEnter the elements for each row (separated by spaces or commas):")
for i in range(m):
    row_input = input(f"Row {i + 1}: ")
    # Clean input, split it, and convert strings to integers
    row = [int(num) for num in row_input.replace(",", " ").split()]

    # Validation check to ensure the user entered the correct number of columns
    while len(row) != n:
        print(f"Error: You must enter exactly {n} numbers.")
        row_input = input(f"Row {i + 1}: ")
        row = [int(num) for num in row_input.replace(",", " ").split()]

    matrix.append(row)

# 3. Calculate and display the sum of each row
print("\n--- Output ---")
for i in range(m):
    # Use the built-in sum() function to add up all numbers in the current row list
    row_sum = sum(matrix[i])
    print(f"Sum of row {i + 1} = {row_sum}")
```
